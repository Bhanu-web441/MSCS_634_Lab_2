### Step 4: Comparative Performance Analysis

#### 1. Performance Observations:
* **KNN Model Trends:** Our testing shows that KNN is highly sensitive to the choice of $K$. At lower $K$ values ($1$ or $5$), the model creates tighter, localized decision boundaries. As $K$ reaches $21$, accuracy tends to plateau or drop because the voting pool expands too far, pulling in majority class votes from neighboring clusters.
* **RNN Model Trends:** After stabilizing the feature dimensions using a `StandardScaler`, the Radius Neighbors Classifier demonstrates high precision when the radius is well-tuned (between $1.5$ and $2.5$). If the radius is too small, points become completely isolated; if it expands to $3.5$, the bounding circle grows too large, pulling in points from multiple classes and causing misclassifications.

#### 2. When to Prefer KNN vs. RNN:
* **Prefer KNN when:** The dataset has uneven or highly variable density distributions, or when you cannot predict the spatial distances between data points. KNN guarantees that every single test instance will find exactly $K$ neighbors and return a prediction, preventing "empty neighborhood" execution crashes.
* **Prefer RNN when:** The data is clean, well-scaled, and evenly distributed, or when your dataset contains notable outliers. Because RNN ignores any data points falling outside its fixed radius constraint, it acts as a natural noise-filtering mechanism that prevents distant anomalies from corrupting local classification decisions.
